```python
df_1['Sales_Target_Achievement_Pct'] = (df_1['Sales_Actual'] / df_1['Sales_Target'] * 100).round(2)
print(df_1[['Store_Id', 'Year_Month', 'Sales_Target', 'Sales_Actual', 'Sales_Target_Achievement_Pct']].head())
```

## Employees

```python
# Xử lý 5 sếp lớn
df_2['Manager_Id'] = df_2['Manager_Id'].fillna('BOD')
df_2['Manager_Name'] = df_2['Manager_Name'].fillna('Board Of Directors')
df_2['Manager_Status'] = df_2['Manager_Status'].fillna('Active')
```

```python
# Trạng thái làm việc & thâm niên
df_2['Is_Active'] = df_2['Exit_Date'].isnull()

report_date = pd.to_datetime('2024-12-31')
df_2['Tenure_Days'] = (df_2['Exit_Date'].fillna(report_date) - df_2['Hire_Date']).dt.days

df_2['Exit_YearMonth'] = df_2['Exit_Date'].dt.to_period('M')
display(df_2.head())
```

```python
max_age = df_2['Age'].max()
print(f'Max age: {max_age}')
```

```python
# Tạo bin tuổi
age_bins = [0,25,40,65]
age_labels = ['Gen Z', 'Millennial', 'Gen X']
df_2['Age_Group'] = pd.cut(df_2['Age'], bins = age_bins, labels = age_labels)
df_2[['Age','Age_Group']]
```
